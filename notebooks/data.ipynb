# Config

```python
import json
import pandas as pd
from pointwise import utils
```

# Amazon Beauty

```python
# JSON lines 형식의 파일 열기
PATH = './data/origin/abeauty.json'
MODE = 'r'
ENCODING = 'utf-8'

kwargs = dict(
    file=PATH, 
    mode=MODE, 
    encoding=ENCODING,
)

with open(**kwargs) as f:
    records = [
        json.loads(line) 
        for line in f
    ]

# reviewerID와 asin만 추출
filtered_records = [
    {
        'reviewerID': r['reviewerID'], 
        'asin': r['asin'],
    } 
    for r in records
]

# DataFrame으로 변환
df = pd.DataFrame(filtered_records)
```

```python
df
```

```python
# Columns
kwargs = dict(
    df=df,
    user_col='reviewerID', 
    item_col='asin', 
)

data = utils.standardizer.main(**kwargs)
```

```python
data
```

```python
# data Description
kwargs = dict(
    df=data,
    percentile=0.9,
)

utils.desc.main(**kwargs)
```

```python
PATH = "./data/abeauty.csv"

kwargs = dict(
    path_or_buf=PATH,
    index=False, 
    header=True,
)

data.to_csv(**kwargs)
```
